```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("titanic")

df.head()
```

```python
data = {
    "Student": ["Aman", "Riya", "Karan", "Neha", "Rohit",
                "Pooja", "Arjun", "Simran", "Vikas", "Anjali",
                "Mohit", "Priya", "Rahul", "Sneha", "Varun"],

    "Gender": ["Male", "Female", "Male", "Female", "Male",
               "Female", "Male", "Female", "Male", "Female",
               "Male", "Female", "Male", "Female", "Male"],

    "Study_Hours": [2, 5, 3, 6, 4, 7, 3, 8, 2, 6,
                    5, 7, 4, 9, 3],

    "Attendance": [65, 90, 72, 95, 80, 92, 70, 98, 60, 88,
                   85, 94, 76, 99, 68],

    "Marks": [45, 78, 55, 88, 65, 92, 58, 95, 40, 82,
              75, 90, 62, 98, 50],

    "Age": [18, 19, 18, 20, 19, 20, 18, 21, 19, 20,
            18, 21, 19, 20, 18],

    "Result": ["Fail", "Pass", "Fail", "Pass", "Pass",
               "Pass", "Fail", "Pass", "Fail", "Pass",
               "Pass", "Pass", "Pass", "Pass", "Fail"]
}

df = pd.DataFrame(data)
df
```

```python
# line plot :- trend/time series
# barplot :- categories comparison
# histplot :- numerical distribution
# scatterplot :- two numerical variable relationship
```

```python
avg_marks = df.groupby("Study_Hours")["Marks"].mean().reset_index()

avg_marks
```

```python
sns.lineplot(
    data=avg_marks,
    x="Study_Hours",
    y="Marks",
    marker="o"
)

plt.xlabel("Study Hours")
plt.ylabel("Average Marks")
plt.title("Average Marks by Study Hours")

plt.show()
```

```python
# barplot() normally numerical variable ka aggregation karta hai
# (categorical feature vs numerical variable ka relationshipsns.

sns.barplot(
    data=df,
    x="Gender",
    y="Marks",
    # errorbar=None            black line shows uncertainity around avg.
)

plt.show()
```

```python
# histogram :- distribution of one numeric value

sns.histplot(
    data=df,
    x="Marks"
)

plt.show()
```
