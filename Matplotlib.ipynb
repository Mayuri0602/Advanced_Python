Data Visualization :- Representation of information in the form of charts, graphs, pictures etc.
                      to understand trends, patterns and outliers

```python
# Matplotlib :- It is a plotting library in Python programming language

# It is used for data visualization introduced by John Hunter in 2002
```

```python
import matplotlib.pyplot as plt
```

```python
# Line Chart :- used to show trends or changes in data over a continuous/ordered sequence, 
#               especially over time


# Common uses: Sales over months, temperature over days, website traffic over time, stock prices,
#               population growth
```

```python
months = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 150, 130, 180]

plt.plot(months, sales)
plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("Monthly Sales Trend")
plt.show()
```

```python
plt.plot(
    months,
    sales,
    color="blue",
    marker="o",
    linewidth=2,
    linestyle="-"
)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid(True)

plt.show()
```

```python
plt.plot(months, sales, marker="o")

for i, value in enumerate(sales):
    plt.text(months[i], value, str(value))

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)

plt.show()
```

```python
# Bar Chart :- used to compare the quantity or value of different categories

# Common uses:- Category-wise sales, product-wise revenue, department-wise employees,city-wise customers
#               comparing marks of students, top 5 products by sales
```

```python
category = ["Electronics", "Clothing", "Furniture", "Beauty", "Sports"]
sales = [50000, 30000, 40000, 20000, 35000]

plt.bar(category, sales,
    color="skyblue",
    width=0.6,
    edgecolor="black",
    linewidth=1)

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")

plt.show()
```

```python
# Histogram :- used to show the distribution of numerical data
#     in simple words, it tells us how frequently values fall within different ranges (bins)
#     (used to understand the distribution and frequency of numerical data)


# Common Uses :- Distribution of age, salary, marks, product prices, distribution of customer spending
```

```python
import matplotlib.pyplot as plt

marks = [45, 52, 55, 61, 63, 67, 72, 75, 78, 82, 85, 90]

plt.hist(marks, bins=5)                        # bins divide data into 5 ranges

plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
# Scatter Plot :- used to understand the relationship or correlation between two numerical variables
#                (used to show the relationship between two numerical variables)

# If dots generally move upward → positive relationship

# If dots generally move downward → negative relationship

# If dots are scattered randomly → little/no relationship
```

```python
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5, 6]
marks = [40, 45, 55, 65, 72, 80]

plt.scatter(hours, marks)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

```python
category = ["Electronics", "Clothing", "Furniture", "Beauty", "Sports"]
sales = [50000, 30000, 40000, 20000, 35000]

plt.pie(
    sales,
    labels=category,
    autopct='%1.1f%%'
)

plt.title('Sales Distribution')
plt.show()
```

```python
# autopct (%1.1f%%) :- used to display % values on each slice of a pie chart 

# % - value ko format karna
# 1.1f - 1 decimal place
# %% - actual % symbol print karna
```

```python
# Subplots :- allow us to display multiple plots in a single figure

# plt.subplot(rows, columns, position)

# ┌───────────┬───────────┐
# │ Plot 1    │ Plot 2    │
# ├───────────┼───────────┤
# │ Plot 3    │ Plot 4    │
# └───────────┴───────────┘


# plt.subplot(2, 2, 1)   # first
# plt.subplot(2, 2, 2)   # second
# plt.subplot(2, 2, 3)   # third
# plt.subplot(2, 2, 4)   # fourth
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 150, 130, 180]
profit = [20, 30, 25, 40]

plt.subplot(1, 2, 1)                 # Line Chart
plt.plot(months, sales)
plt.title("Sales")

plt.subplot(1, 2, 2)                # Bar chart
plt.bar(months, profit)
plt.title("Profit")

plt.show()
```
